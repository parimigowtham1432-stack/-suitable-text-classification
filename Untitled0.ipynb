```python

# ==============================================================
# CLA-2 TASK 2
# MULTI-CLASS TEXT CLASSIFICATION USING MACHINE LEARNING
# Algorithms: Decision Tree, KNN, SVM, Naive Bayes
# Dataset: 20 Newsgroups
# ==============================================================

# STEP 1: INSTALL LIBRARIES
!pip install -q scikit-learn pandas numpy matplotlib seaborn

# STEP 2: IMPORT LIBRARIES
import os
import json
import shutil
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from google.colab import files

warnings.filterwarnings("ignore")

# CREATE OUTPUT FOLDER
OUTPUT_DIR = "/content/CLA2_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 65)
print("CLA-2: TEXT CLASSIFICATION COMPARATIVE STUDY")
print("=" * 65)


# STEP 3: LOAD DATASET

categories = [
    "comp.graphics",
    "rec.sport.baseball",
    "sci.med",
    "talk.politics.misc"
]

print("\nDownloading 20 Newsgroups dataset...")

dataset = fetch_20newsgroups(
    subset="all",
    categories=categories,
    remove=("headers", "footers", "quotes"),
    shuffle=True,
    random_state=42
)

X = dataset.data
y = dataset.target

class_names = dataset.target_names

print("\nDataset loaded successfully!")
print("Total documents:", len(X))
print("Number of classes:", len(class_names))
print("Classes:", class_names)


# STEP 4: CLASS DISTRIBUTION

class_counts = np.bincount(y, minlength=len(class_names))

distribution = pd.DataFrame({
    "Category": class_names,
    "Number of Documents": class_counts
})

print("\nCLASS DISTRIBUTION")
print(distribution.to_string(index=False))

distribution.to_csv(
    os.path.join(OUTPUT_DIR, "class_distribution.csv"),
    index=False
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=distribution,
    x="Category",
    y="Number of Documents"
)

plt.title("Class Distribution of 20 Newsgroups Dataset")
plt.xlabel("News Categories")
plt.ylabel("Number of Documents")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "class_distribution.png"),
    dpi=300
)
plt.show()


# STEP 5: TRAIN-TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nDATASET SPLITTING")
print("Training documents:", len(X_train))
print("Testing documents:", len(X_test))


# STEP 6: TEXT PREPROCESSING AND TF-IDF

print("\nPerforming text preprocessing and TF-IDF...")

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=30000,
    min_df=2,
    sublinear_tf=True
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF completed!")
print("Training feature shape:", X_train_tfidf.shape)
print("Testing feature shape:", X_test_tfidf.shape)
print("Total features:", len(tfidf.get_feature_names_out()))


# STEP 7: DEFINE FOUR MACHINE LEARNING MODELS

models = {
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        metric="cosine",
        algorithm="brute"
    ),

    "SVM": LinearSVC(
        C=1.0,
        random_state=42
    ),

    "Naive Bayes": MultinomialNB()
}


# STEP 8: TRAIN MODELS AND EVALUATE

results = []
predictions = {}
trained_models = {}

print("\nTRAINING AND EVALUATING MODELS")

for name, model in models.items():

    print("\n" + "=" * 50)
    print("Training:", name)

    model.fit(X_train_tfidf, y_train)

    y_pred = model.predict(X_test_tfidf)

    predictions[name] = y_pred
    trained_models[name] = model

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    # SAVE CLASSIFICATION REPORT
    report = classification_report(
        y_test,
        y_pred,
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    report_df = pd.DataFrame(report).transpose()

    report_df.to_csv(
        os.path.join(
            OUTPUT_DIR,
            name.lower().replace(" ", "_") + "_report.csv"
        )
    )


# STEP 9: COMPARISON TABLE

results_df = pd.DataFrame(results)

print("\n" + "=" * 65)
print("FINAL MODEL COMPARISON")
print("=" * 65)

print(results_df.round(4).to_string(index=False))

results_df.to_csv(
    os.path.join(OUTPUT_DIR, "model_comparison.csv"),
    index=False
)


# STEP 10: PERFORMANCE COMPARISON GRAPH

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

plt.figure(figsize=(12, 6))

results_df.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Comparison of Four Text Classification Algorithms")
plt.xlabel("Machine Learning Models")
plt.ylabel("Performance Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.legend(loc="lower right")
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "model_comparison.png"),
    dpi=300
)
plt.show()


# STEP 11: CONFUSION MATRICES

print("\nGENERATING CONFUSION MATRICES")

for name, y_pred in predictions.items():

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=np.arange(len(class_names))
    )

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names
    )

    plt.title(name + " - Confusion Matrix")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.xticks(rotation=25)
    plt.yticks(rotation=0)
    plt.tight_layout()

    filename = (
        name.lower().replace(" ", "_")
        + "_confusion_matrix.png"
    )

    plt.savefig(
        os.path.join(OUTPUT_DIR, filename),
        dpi=300
    )

    plt.show()


# STEP 12: CLASSIFICATION REPORTS

print("\nDETAILED CLASSIFICATION REPORTS")

for name, y_pred in predictions.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=class_names,
            zero_division=0
        )
    )


# STEP 13: WHAT-IF EXPERIMENT
# Change KNN K values: 3, 5, 7, 9

print("\nWHAT-IF EXPERIMENT: CHANGING KNN K VALUE")

knn_results = []

for k in [3, 5, 7, 9]:

    print("\nTesting K =", k)

    knn_model = KNeighborsClassifier(
        n_neighbors=k,
        metric="cosine",
        algorithm="brute"
    )

    knn_model.fit(X_train_tfidf, y_train)

    knn_pred = knn_model.predict(X_test_tfidf)

    knn_results.append({
        "K Value": k,
        "Accuracy": accuracy_score(y_test, knn_pred),
        "Precision": precision_score(
            y_test, knn_pred,
            average="macro",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, knn_pred,
            average="macro",
            zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, knn_pred,
            average="macro",
            zero_division=0
        )
    })

knn_df = pd.DataFrame(knn_results)

print("\nKNN WHAT-IF RESULTS")
print(knn_df.round(4).to_string(index=False))

knn_df.to_csv(
    os.path.join(OUTPUT_DIR, "knn_what_if_results.csv"),
    index=False
)


# STEP 14: WHAT-IF EXPERIMENT GRAPH

plt.figure(figsize=(10, 5))

knn_df.set_index("K Value")[metrics].plot(
    marker="o",
    figsize=(10, 5)
)

plt.title("Effect of K Value on KNN Performance")
plt.xlabel("K Value")
plt.ylabel("Performance Score")
plt.ylim(0, 1)
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "knn_what_if_experiment.png"),
    dpi=300
)
plt.show()


# STEP 15: IDENTIFY BEST MODEL

best_model = results_df.loc[
    results_df["F1 Score"].idxmax()
]

print("\n" + "=" * 65)
print("BEST MODEL ANALYSIS")
print("=" * 65)

print("Best Model:", best_model["Model"])
print("Accuracy:", round(best_model["Accuracy"], 4))
print("Precision:", round(best_model["Precision"], 4))
print("Recall:", round(best_model["Recall"], 4))
print("F1 Score:", round(best_model["F1 Score"], 4))

best_knn = knn_df.loc[knn_df["F1 Score"].idxmax()]

print("\nBest KNN K Value:", int(best_knn["K Value"]))
print("Best KNN F1 Score:", round(best_knn["F1 Score"], 4))


# STEP 16: SAVE EXPERIMENT SUMMARY

summary = {
    "Project": "Multi-Class News Text Classification",
    "Dataset": "20 Newsgroups",
    "Categories": class_names,
    "Total Documents": len(X),
    "Training Documents": len(X_train),
    "Testing Documents": len(X_test),
    "TF-IDF Features": len(tfidf.get_feature_names_out()),
    "Evaluation": "Macro Precision, Recall and F1 Score",
    "Best Model": str(best_model["Model"]),
    "Best Model F1 Score": float(best_model["F1 Score"]),
    "Best KNN K Value": int(best_knn["K Value"])
}

with open(
    os.path.join(OUTPUT_DIR, "experiment_summary.json"),
    "w"
) as f:
    json.dump(summary, f, indent=4)


# STEP 17: SAVE TEST PREDICTIONS

prediction_df = pd.DataFrame({
    "Actual Category": [
        class_names[i] for i in y_test
    ]
})

for name, y_pred in predictions.items():

    column_name = name + " Prediction"

    prediction_df[column_name] = [
        class_names[i] for i in y_pred
    ]

prediction_df.to_csv(
    os.path.join(OUTPUT_DIR, "test_predictions.csv"),
    index=False
)


# STEP 18: ZIP ALL OUTPUT FILES

zip_path = shutil.make_archive(
    "/content/CLA2_Experiment_Results",
    "zip",
    OUTPUT_DIR
)

print("\n" + "=" * 65)
print("EXPERIMENT COMPLETED SUCCESSFULLY!")
print("=" * 65)

print("\nGenerated output files:")

for filename in sorted(os.listdir(OUTPUT_DIR)):
    print("-", filename)

print("\nZIP file created:", zip_path)

print("\nDownloading all results...")

files.download(zip_path)

print("\nPLEASE SAVE YOUR RESULTS AND USE ACTUAL METRICS")
print("IN YOUR ASSIGNMENT REPORT.")

```
